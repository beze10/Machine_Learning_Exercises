```python
import math

def logistic_probability(beta_0, beta_1, X):
    # Calculate the exponent part
    exponent = beta_0 + beta_1 * X
    # Calculate the probability using the logistic function
    p_x = math.exp(exponent) / (1 + math.exp(exponent))
    return p_x

# Example usage
beta_0 = -10.6513
beta_1 = 0.0055
X = 3000
probability = logistic_probability(beta_0, beta_1, X)
print(f"The probability p(X) is: {probability}")
```

```python
import numpy as np
import matplotlib.pyplot as plt


# X values for the plot
X_values = np.linspace(0, 4000, 100)  # Generate 100 points between 0 and 2000
p_values = [logistic_probability(beta_0, beta_1, X) for X in X_values]

# Plot p(X) vs X
plt.figure(figsize=(8, 6))
plt.plot(X_values, p_values, label='p(X)', color='b')
plt.title('Logistic Probability p(X) vs X')
plt.xlabel('X')
plt.ylabel('p(X)')
plt.grid(True)
plt.legend()
plt.show()
```
